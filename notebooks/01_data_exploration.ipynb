# Cricket Data Exploration

This notebook performs exploratory data analysis (EDA) on T20 cricket data.

## Objectives:
- Load and explore cricket match data
- Understand data structure and quality
- Identify patterns and insights
- Generate visualizations

```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
from pathlib import Path
import warnings

warnings.filterwarnings('ignore')

# Set style
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)

# Add src to path for imports
import sys
sys.path.append(str(Path.cwd().parent / 'src'))

from etl.data_loader import CricketDataLoader
from etl.data_processor import DataProcessor

print("Libraries imported successfully!")
```

## 1. Load Data

Load cricket match data using the data loader. This will integrate with CricPy API.

```python
# Initialize data loader
loader = CricketDataLoader()

# TODO: Load data from CricPy API
# For now, this is a placeholder structure
# match_data = loader.load_tournament_data('IPL', '2023')

# Sample data structure (replace with actual API data)
match_data = pd.DataFrame()

print(f"Data shape: {match_data.shape}")
match_data.head() if not match_data.empty else "No data loaded yet - integrate CricPy API"
```

## 2. Data Overview

```python
if not match_data.empty:
    print("Dataset Info:")
    print(match_data.info())
    print("\nBasic Statistics:")
    print(match_data.describe())
    print("\nMissing Values:")
    print(match_data.isnull().sum())
else:
    print("Awaiting data from CricPy API integration")
```

## 3. Match Statistics Analysis

```python
# Distribution of total runs per match
if not match_data.empty and 'total_runs' in match_data.columns:
    fig = px.histogram(match_data, x='total_runs', nbins=30,
                      title='Distribution of Total Runs per Match')
    fig.show()
else:
    print("Visualization will be available after data integration")
```

```python
# Wickets distribution
if not match_data.empty and 'wickets' in match_data.columns:
    fig = px.box(match_data, y='wickets',
                title='Wickets Distribution in T20 Matches')
    fig.show()
```

## 4. Team Performance Analysis

```python
# Win percentage by team
if not match_data.empty and 'team' in match_data.columns and 'result' in match_data.columns:
    team_wins = match_data.groupby('team')['result'].value_counts(normalize=True).unstack()
    team_wins.plot(kind='bar', stacked=True, figsize=(14, 6))
    plt.title('Win Percentage by Team')
    plt.ylabel('Percentage')
    plt.xlabel('Team')
    plt.legend(title='Result')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

## 5. Player Performance Analysis

```python
# Top scorers
if not match_data.empty and 'player' in match_data.columns and 'runs' in match_data.columns:
    top_scorers = match_data.groupby('player')['runs'].sum().sort_values(ascending=False).head(10)
    
    fig = px.bar(x=top_scorers.values, y=top_scorers.index,
                orientation='h',
                title='Top 10 Run Scorers',
                labels={'x': 'Total Runs', 'y': 'Player'})
    fig.show()
```

## 6. Venue Analysis

```python
# Average scores by venue
if not match_data.empty and 'venue' in match_data.columns and 'total_runs' in match_data.columns:
    venue_avg = match_data.groupby('venue')['total_runs'].mean().sort_values(ascending=False).head(10)
    
    plt.figure(figsize=(12, 6))
    venue_avg.plot(kind='barh')
    plt.title('Average Runs by Venue (Top 10)')
    plt.xlabel('Average Runs')
    plt.ylabel('Venue')
    plt.tight_layout()
    plt.show()
```

## 7. Time-based Analysis

```python
# Trends over time
if not match_data.empty and 'date' in match_data.columns:
    match_data['date'] = pd.to_datetime(match_data['date'])
    
    # Runs trend over time
    if 'total_runs' in match_data.columns:
        time_trend = match_data.groupby(match_data['date'].dt.to_period('M'))['total_runs'].mean()
        
        plt.figure(figsize=(14, 6))
        time_trend.plot()
        plt.title('Average Runs Trend Over Time')
        plt.xlabel('Date')
        plt.ylabel('Average Runs')
        plt.grid(True)
        plt.tight_layout()
        plt.show()
```

## 8. Correlation Analysis

```python
# Correlation heatmap
if not match_data.empty:
    numeric_cols = match_data.select_dtypes(include=[np.number]).columns
    
    if len(numeric_cols) > 1:
        correlation = match_data[numeric_cols].corr()
        
        plt.figure(figsize=(12, 8))
        sns.heatmap(correlation, annot=True, cmap='coolwarm', center=0)
        plt.title('Feature Correlation Heatmap')
        plt.tight_layout()
        plt.show()
```

## 9. Key Insights Summary

```python
# Generate summary statistics and insights
if not match_data.empty:
    insights = {
        'Total Matches': len(match_data),
        'Date Range': f"{match_data['date'].min()} to {match_data['date'].max()}" if 'date' in match_data.columns else 'N/A',
        'Unique Teams': match_data['team'].nunique() if 'team' in match_data.columns else 'N/A',
        'Unique Venues': match_data['venue'].nunique() if 'venue' in match_data.columns else 'N/A',
    }
    
    print("=" * 50)
    print("KEY INSIGHTS")
    print("=" * 50)
    for key, value in insights.items():
        print(f"{key}: {value}")
else:
    print("\n" + "="*50)
    print("NOTE: This notebook is ready for data integration.")
    print("Once CricPy API is connected, all visualizations will populate automatically.")
    print("="*50)
```
